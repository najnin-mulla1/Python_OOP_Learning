```python
class ATM:
    def __init__(self, balance=0):
        self.balance = balance
        self.pin = None

        print(f""" Welcome to your account! select following options for transactions:
        1. Check available balance
        2. Create PIN
        3. Verify PIN
        4. Credit account
        5. Debit account
        """)
        self.create_pin()

    def available_balance(self):
        print(f"Your current balance is: {self.balance}")
        return self.balance

    def create_pin(self):
        pin = input("Enter a new PIN: ")
        self.pin = pin
        print("PIN is created successfully.\n")

    def verify_pin(self,pin):
        #entered_pin = input("Enter your PIN to verify: ")
        # if entered_pin == self.pin:
        #     print("PIN verified successfully!\n")
        #     return True
        # else:
        #     print("Incorrect PIN! Access denied.\n")
        #     return False
        return self.pin == pin

    def credit(self, amount):
        self.balance += amount
        print(f"Successfully credited {amount}. New balance: {self.balance}\n")
        return self.balance

    def debit(self, amount):
        if amount > self.balance:
            print("Insufficient funds!\n")
            return "Insufficient funds"
        self.balance -= amount
        print(f"Successfully debited {amount}. New balance: {self.balance}\n")
        return self.balance 

SBI = ATM()

# SBI.verify_pin(input("Enter your PIN to verify: "))

# SBI.credit(float(input("Enter amount to credit:")))

# HDFC = ATM()
# HDFC.credit(10000)

# SBI.available_balance()

# SBI.debit(float(input("Enter amount to debit")))

# HDFC.available_balance()
```
